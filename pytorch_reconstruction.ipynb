```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import pandas as pd

# put the imports here....
# Load numerical data, skipping the first header row
v = np.loadtxt('v.csv', delimiter=',', skiprows=0)
s_diag = np.loadtxt('s.csv', delimiter=',', skiprows=0)
u = np.loadtxt('u.csv', delimiter=',', skiprows=0)
s = np.diag(s_diag)
```

```python
v_tensor = torch.from_numpy(v)
u_tensor = torch.from_numpy(u)
s_tensor = torch.from_numpy(s)
```

```python
u_s = torch.mm(u_tensor, s_tensor)
reconsturcted = torch.mm(u_s, v_tensor)

plt.subplots(figsize=(2, 2)) # choose size of plot
plt.axis('off') # turn off axis numbering
plt.imshow(reconsturcted, cmap=plt.cm.gray ) # plot the image
```

```python
k = 30

u_reduced = u_tensor[:, :k]
s_reduced = s_tensor[:k, :k]
v_reduced = v_tensor[:k, :]

u_s = torch.mm(u_reduced, s_reduced)
reconsturcted = torch.mm(u_s, v_reduced)

plt.subplots(figsize=(2, 2)) # choose size of plot
plt.axis('off') # turn off axis numbering
plt.imshow(reconsturcted, cmap=plt.cm.gray )
```
